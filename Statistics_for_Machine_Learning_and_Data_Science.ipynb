```python
# ==========================================
# EXPERIMENT 3
# Correlation, Distance Measures &
# Data Preprocessing
# ==========================================

from sklearn.preprocessing import StandardScaler

# 1. Medical columns
medical_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

# 2. Create copy
df_preprocessed = df.copy()

# 3. Replace invalid zero values
df_preprocessed[medical_columns] = (
    df_preprocessed[medical_columns].replace(0, np.nan)
)

# 4. Median imputation
for column in medical_columns:
    df_preprocessed[column] = df_preprocessed[column].fillna(
        df_preprocessed[column].median()
    )

# 5. Correlation matrix
correlation_matrix = df_preprocessed.corr(numeric_only=True)

print("Correlation Matrix:")
display(correlation_matrix)

# 6. Heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Heatmap")
plt.show()

# 7. Features for distance calculation
features = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age"
]

# 8. Select two observations
x = df_preprocessed.loc[0, features].values
y = df_preprocessed.loc[1, features].values

# 9. Distance calculations
euclidean_distance = np.sqrt(np.sum((x - y) ** 2))

manhattan_distance = np.sum(np.abs(x - y))

cosine_similarity = np.dot(x, y) / (
    np.linalg.norm(x) * np.linalg.norm(y)
)

cosine_distance = 1 - cosine_similarity

print("Euclidean Distance:", euclidean_distance)
print("Manhattan Distance:", manhattan_distance)
print("Cosine Distance:", cosine_distance)

# 10. Standardization
scaler = StandardScaler()

df_scaled = df_preprocessed.copy()

df_scaled[features] = scaler.fit_transform(
    df_preprocessed[features]
)

# 11. Compare before and after
print("\nBefore Standardization:")
display(df_preprocessed[features].head())

print("\nAfter Standardization:")
display(df_scaled[features].head())
```

```python
#Point Estimation — Sample Proportion
diabetic_proportion = df["Outcome"].mean()

print("Estimated population proportion of diabetic patients:",
      diabetic_proportion)
```

```python
#95% Confidence Interval for Mean Glucose
from scipy import stats

glucose = df["Glucose"]

n = len(glucose)
mean = glucose.mean()
std = glucose.std()

confidence = 0.95

t_critical = stats.t.ppf(
    1 - (1-confidence)/2,
    n-1
)

margin_error = t_critical * std / np.sqrt(n)

lower = mean - margin_error
upper = mean + margin_error

print("Mean Glucose:", mean)
print("95% Confidence Interval:")
print("Lower:", lower)
print("Upper:", upper)
```

```python
# One-Sample t-Test
# Test whether mean Glucose is different from 120.

t_stat, p_value = stats.ttest_1samp(
    df["Glucose"],
    120
)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
#Decision for One-Sample t-Test
alpha = 0.05

if p_value < alpha:
    print("Reject H0")
else:
    print("Fail to reject H0")
```

```python
#Two-Sample t-Test — BMI
bmi_non_diabetic = df[df["Outcome"] == 0]["BMI"]
bmi_diabetic = df[df["Outcome"] == 1]["BMI"]

t_stat_bmi, p_value_bmi = stats.ttest_ind(
    bmi_non_diabetic,
    bmi_diabetic,
    equal_var=False
)

print("t-statistic:", t_stat_bmi)
print("p-value:", p_value_bmi)
```

```python
#Decision for Two-Sample t-Test
if p_value_bmi < 0.05:
    print("Reject H0")
    print("BMI differs significantly between the two groups.")
else:
    print("Fail to reject H0")
    print("No significant difference in BMI was established.")
```

```python
# Create Age Groups for Chi-Square Test
# Since the Pima dataset contains only female patients, there is no Gender variable to use for the example in the experiment. We can use Age Group vs Outcome instead.
df["AgeGroup"] = pd.cut(
    df["Age"],
    bins=[20, 30, 40, 50, 100],
    labels=["21-30", "31-40", "41-50", "51+"]
)
```

```python
#Contingency Table
contingency_table = pd.crosstab(
    df["AgeGroup"],
    df["Outcome"]
)

print(contingency_table)
```

```python
#Chi-Square Test
chi2_stat, chi2_p, dof, expected = stats.chi2_contingency(
    contingency_table
)

print("Chi-square statistic:", chi2_stat)
print("p-value:", chi2_p)
print("Degrees of freedom:", dof)
```

```python
#. Decision for Chi-Square Test
if chi2_p < 0.05:
    print("Reject H0")
    print("There is a significant association between Age Group and Outcome.")
else:
    print("Fail to reject H0")
    print("There is no significant association established between Age Group and Outcome.")
```

```python
# ============================================================
# EXPERIMENT 5
# Implement Resampling Techniques and Confidence Interval
# Estimation for Statistical Decision Making
# ============================================================

# -----------------------------
# IMPORT REQUIRED LIBRARIES
# -----------------------------

import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns


# ============================================================
# 1. LOAD DATASET
# ============================================================

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/diabetes.csv')

print("First 5 Records:")
print(df.head())

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)


# ============================================================
# 2. SELECT NUMERICAL VARIABLE
# ============================================================

# We select Glucose for bootstrap analysis

glucose = df["Glucose"].dropna()

print("\n==============================")
print("SELECTED VARIABLE: GLUCOSE")
print("==============================")

print("Number of observations:", len(glucose))
print("Original Sample Mean:", glucose.mean())
print("Original Sample Standard Deviation:", glucose.std())


# ============================================================
# 3. BOOTSTRAP RESAMPLING
# ============================================================

np.random.seed(42)

n_bootstrap = 10000

bootstrap_means = []

for i in range(n_bootstrap):

    # Create bootstrap sample with replacement
    sample = np.random.choice(
        glucose,
        size=len(glucose),
        replace=True
    )

    # Calculate mean of bootstrap sample
    bootstrap_mean = sample.mean()

    bootstrap_means.append(bootstrap_mean)

# Convert list to NumPy array
bootstrap_means = np.array(bootstrap_means)

print("\n==============================")
print("BOOTSTRAP RESAMPLING")
print("==============================")

print("Number of Bootstrap Samples:",
      n_bootstrap)

print("Original Sample Mean:",
      glucose.mean())

print("Mean of Bootstrap Distribution:",
      bootstrap_means.mean())


# ============================================================
# 4. BOOTSTRAP 95% CONFIDENCE INTERVAL
# ============================================================

lower = np.percentile(
    bootstrap_means,
    2.5
)

upper = np.percentile(
    bootstrap_means,
    97.5
)

print("\n==============================")
print("95% BOOTSTRAP CONFIDENCE INTERVAL")
print("==============================")

print("Lower Bound:", lower)
print("Upper Bound:", upper)

print(
    f"95% Bootstrap Confidence Interval: "
    f"({lower:.2f}, {upper:.2f})"
)


# ============================================================
# 5. BOOTSTRAP DISTRIBUTION VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    bootstrap_means,
    bins=50,
    kde=True
)

plt.axvline(
    glucose.mean(),
    linestyle="--",
    linewidth=2,
    label="Original Mean"
)

plt.axvline(
    lower,
    linestyle="--",
    linewidth=2,
    label="2.5% CI"
)

plt.axvline(
    upper,
    linestyle="--",
    linewidth=2,
    label="97.5% CI"
)

plt.title(
    "Bootstrap Distribution of Mean Glucose"
)

plt.xlabel("Bootstrap Mean Glucose")
plt.ylabel("Frequency")

plt.legend()

plt.show()


# ============================================================
# 6. DIVIDE DATA INTO TWO GROUPS
# ============================================================

# Outcome:
# 0 = Non-Diabetic
# 1 = Diabetic

diabetic = df[
    df["Outcome"] == 1
]["Glucose"].dropna()

non_diabetic = df[
    df["Outcome"] == 0
]["Glucose"].dropna()

print("\n==============================")
print("GROUP INFORMATION")
print("==============================")

print(
    "Number of Diabetic Patients:",
    len(diabetic)
)

print(
    "Number of Non-Diabetic Patients:",
    len(non_diabetic)
)

print(
    "\nMean Glucose - Diabetic:",
    diabetic.mean()
)

print(
    "Mean Glucose - Non-Diabetic:",
    non_diabetic.mean()
)


# ============================================================
# 7. CALCULATE OBSERVED DIFFERENCE
# ============================================================

observed_difference = (
    diabetic.mean()
    - non_diabetic.mean()
)

print("\n==============================")
print("OBSERVED DIFFERENCE")
print("==============================")

print(
    "Observed Difference in Mean Glucose:",
    observed_difference
)


# ============================================================
# 8. PERMUTATION TEST
# ============================================================

np.random.seed(42)

n_permutations = 10000

# Combine both groups
combined = np.concatenate(
    [
        diabetic.values,
        non_diabetic.values
    ]
)

# Number of observations in diabetic group
n_diabetic = len(diabetic)

permutation_differences = []

for i in range(n_permutations):

    # Randomly shuffle the combined data
    shuffled = np.random.permutation(
        combined
    )

    # Divide shuffled data into two groups
    shuffled_diabetic = shuffled[
        :n_diabetic
    ]

    shuffled_non_diabetic = shuffled[
        n_diabetic:
    ]

    # Calculate difference in means
    difference = (
        shuffled_diabetic.mean()
        - shuffled_non_diabetic.mean()
    )

    permutation_differences.append(
        difference
    )

# Convert to NumPy array
permutation_differences = np.array(
    permutation_differences
)

print("\n==============================")
print("PERMUTATION TEST")
print("==============================")

print(
    "Number of Permutations:",
    n_permutations
)

print("Permutation Test Completed.")


# ============================================================
# 9. CALCULATE PERMUTATION P-VALUE
# ============================================================

p_value = np.mean(
    np.abs(permutation_differences)
    >= abs(observed_difference)
)

alpha = 0.05

print("\n==============================")
print("PERMUTATION TEST RESULTS")
print("==============================")

print(
    "Observed Difference:",
    observed_difference
)

print(
    "P-value:",
    p_value
)

print(
    "Significance Level (alpha):",
    alpha
)


# ============================================================
# 10. HYPOTHESIS DECISION
# ============================================================

print("\n==============================")
print("HYPOTHESIS TEST DECISION")
print("==============================")

print(
    "H0: There is no significant difference "
    "in mean glucose between diabetic and "
    "non-diabetic groups."
)

print(
    "H1: There is a significant difference "
    "in mean glucose between diabetic and "
    "non-diabetic groups."
)

if p_value < alpha:

    print("\nDecision: Reject H0")

    print(
        "Conclusion: There is sufficient "
        "statistical evidence that the mean "
        "glucose levels differ between "
        "diabetic and non-diabetic groups."
    )

else:

    print("\nDecision: Fail to Reject H0")

    print(
        "Conclusion: There is insufficient "
        "statistical evidence to conclude "
        "that the mean glucose levels differ "
        "between the two groups."
    )


# ============================================================
# 11. PERMUTATION DISTRIBUTION VISUALIZATION
# ============================================================

plt.figure(figsize=(10, 6))

sns.histplot(
    permutation_differences,
    bins=50,
    kde=True
)

plt.axvline(
    observed_difference,
    linestyle="--",
    linewidth=2,
    label="Observed Difference"
)

plt.axvline(
    -observed_difference,
    linestyle="--",
    linewidth=2
)

plt.title(
    "Permutation Distribution of Mean Glucose Difference"
)

plt.xlabel(
    "Difference in Mean Glucose"
)

plt.ylabel("Frequency")

plt.legend()

plt.show()


# ============================================================
# 12. FINAL SUMMARY
# ============================================================

print("\n")
print("====================================================")
print("             EXPERIMENT 5 - FINAL RESULTS")
print("====================================================")

print("\nBOOTSTRAP RESULTS")
print("--------------------------------------------")

print(
    f"Original Mean Glucose: "
    f"{glucose.mean():.2f}"
)

print(
    f"Bootstrap Mean: "
    f"{bootstrap_means.mean():.2f}"
)

print(
    f"95% Bootstrap Confidence Interval: "
    f"({lower:.2f}, {upper:.2f})"
)

print("\nPERMUTATION TEST RESULTS")
print("--------------------------------------------")

print(
    f"Mean Glucose - Diabetic: "
    f"{diabetic.mean():.2f}"
)

print(
    f"Mean Glucose - Non-Diabetic: "
    f"{non_diabetic.mean():.2f}"
)

print(
    f"Observed Difference: "
    f"{observed_difference:.2f}"
)

print(
    f"Permutation P-value: "
    f"{p_value:.6f}"
)

print(
    f"Significance Level: "
    f"{alpha}"
)

if p_value < alpha:
    print("Final Decision: Reject H0")
else:
    print("Final Decision: Fail to Reject H0")

print("====================================================")
print("                 EXPERIMENT COMPLETED")
print("====================================================")
```
